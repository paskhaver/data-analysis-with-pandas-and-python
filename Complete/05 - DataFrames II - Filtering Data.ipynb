## Filter A DataFrame  Based On A Condition
- Pandas needs a Boolean `Series` to perform a filter.
- Pass the Boolean `Series` inside square brackets after the `DataFrame`.

```python
employees = pd.read_csv("employees.csv").astype(
    {"Senior Management": bool, "Gender": "category", "Team": "category"}
)

employees["Start Date"] = pd.to_datetime(employees["Start Date"], format="%m/%d/%Y")
employees["Last Login Time"] = pd.to_datetime(
    employees["Last Login Time"], format="%H:%M %p"
).dt.time

employees.head()
```

```python
5 == 6
5 != 10
```

```python
employees[employees["Gender"] == "Male"]
employees[employees["Gender"] != "Male"]
employees[employees["Team"] == "Finance"]
```

- Assign the Boolean `Series` to a variable for a simpler syntax.

```python
on_finance_team = employees["Team"] == "Finance"
employees[on_finance_team]
```

- An existing Boolean column can serve as a filter.

```python
employees[employees["Senior Management"]]
```

- We can generate a Boolean `Series` using a wide variety of operations (equality, inequality, less than, greater than, inclusion, etc)

```python
employees[employees["Salary"] > 110000]

employees[employees["Bonus %"] < 1.5]
```

```python
employees[employees["Start Date"] < "1985-01-01"]
```

```python
import datetime as dt
```

```python
employees[employees["Last Login Time"] < dt.time(12, 0, 0)]
```

## Filter with More than One Condition (AND)

```python
employees = pd.read_csv("employees.csv").astype(
    {"Senior Management": bool, "Gender": "category", "Team": "category"}
)

employees["Start Date"] = pd.to_datetime(employees["Start Date"], format="%m/%d/%Y")
employees["Last Login Time"] = pd.to_datetime(
    employees["Last Login Time"], format="%H:%M %p"
).dt.time

employees.head()
```

- Add the `&` operator in between two Boolean `Series` to filter by multiple conditions.
- Add parentheses around the individual `Series` to avoid errors/confusion.

```python
# female employees who work in Marketing
employees[(employees["Gender"] == "Female") & (employees["Team"] == "Marketing")]
```

- We can assign the `Series` to variables to make the syntax more readable.
- Variables also remove the need for parentheses.

```python
is_female = employees["Gender"] == "Female"
is_in_marketing = employees["Team"] == "Marketing"
salary_over_100k = employees["Salary"] > 100000

employees[is_female & is_in_marketing & salary_over_100k].head()
```

## Filter with More than One Condition (OR)
- Use the `|` operator in between two Boolean `Series` to filter by *either* condition.

```python
employees = pd.read_csv("employees.csv").astype(
    {"Senior Management": bool, "Gender": "category", "Team": "category"}
)

employees["Start Date"] = pd.to_datetime(employees["Start Date"], format="%m/%d/%Y")
employees["Last Login Time"] = pd.to_datetime(
    employees["Last Login Time"], format="%H:%M %p"
).dt.time

employees.head()
```

- Let's target employees who are either senior management OR who started before 1990-01-01.

```python
is_senior_management = employees["Senior Management"]
started_in_80s = employees["Start Date"] < "1990-01-01"

employees[is_senior_management | started_in_80s]
```

- Let's look for employees named Robert who work in Client Services OR have a start date after 2016-06-01.

```python
is_robert = employees["First Name"] == "Robert"
is_in_client_services = employees["Team"] == "Client Services"
start_date_after_june_2016 = employees["Start Date"] > "2016-06-01"
```

```python
employees[(is_robert & is_in_client_services) | start_date_after_june_2016]

employees[is_robert & (is_in_client_services | start_date_after_june_2016)]

# employees[is_robert & is_in_client_services | start_date_after_june_2016]
```

## The isin Method
- The `isin` method accepts a collection object like a list, tuple, or `Series`.
- The method returns True for a row if its value is found within the collection.

```python
employees = pd.read_csv("employees.csv").astype(
    {"Senior Management": bool, "Gender": "category", "Team": "category"}
)

employees["Start Date"] = pd.to_datetime(employees["Start Date"], format="%m/%d/%Y")
employees["Last Login Time"] = pd.to_datetime(
    employees["Last Login Time"], format="%H:%M %p"
).dt.time

employees.head()
```

- Let's say we're targeting employees whose team is Legal, Sales, or Product.

```python
legal_team = employees["Team"] == "Legal"
sales_team = employees["Team"] == "Sales"
product_team = employees["Team"] == "Product"

employees[legal_team | sales_team | product_team]
```

```python
target_teams = employees["Team"].isin(["Legal", "Sales", "Product"])
employees[target_teams]
```

## The isnull and notnull Methods
- The `isnull` method returns True for `NaN` values in a `Series`.
- The `notnull` method returns True for present values in a `Series`.
- The `isna` method is an alias for `isnull`.
- The `notna` method is an alias for `notnull`.

```python
employees = pd.read_csv("employees.csv").astype(
    {"Senior Management": bool, "Gender": "category", "Team": "category"}
)

employees["Start Date"] = pd.to_datetime(employees["Start Date"], format="%m/%d/%Y")
employees["Last Login Time"] = pd.to_datetime(
    employees["Last Login Time"], format="%H:%M %p"
).dt.time

employees.head()
```

```python
employees[employees["Team"].isnull()]
employees[employees["Team"].isna()]
```

```python
employees[employees["Team"].notnull()]
employees[employees["Team"].notna()]
```

```python
employees[employees["First Name"].isnull() & employees["Team"].notnull()]
```

## The between Method
- The `between` method returns True if a `Series` value is found within its range.

```python
employees = pd.read_csv("employees.csv").astype(
    {"Senior Management": bool, "Gender": "category", "Team": "category"}
)

employees["Start Date"] = pd.to_datetime(employees["Start Date"], format="%m/%d/%Y")
employees["Last Login Time"] = pd.to_datetime(
    employees["Last Login Time"], format="%H:%M %p"
).dt.time

employees.head()
```

```python
import datetime as dt
```

```python
employees[employees["Salary"].between(60000, 70000)]
employees[employees["Bonus %"].between(2.0, 5.0)]
employees[employees["Start Date"].between("1991-01-01", "1992-01-01")]
employees[employees["Last Login Time"].between(dt.time(8, 30), dt.time(12, 0))]
```

## The duplicated Method
- The `duplicated` method returns True if a `Series` value is a duplicate.
- Pandas will mark one occurrence of a repeated value as a non-duplicate.
- Use the `keep` parameter to designate whether to consider the first or last occurrence of a repeated value as the non-duplicate.
- Pass `False` to the `keep` parameter to mark all occurrences of repeated values as duplicates.

```python
employees = pd.read_csv("employees.csv").astype(
    {"Senior Management": bool, "Gender": "category", "Team": "category"}
)

employees["Start Date"] = pd.to_datetime(employees["Start Date"], format="%m/%d/%Y")
employees["Last Login Time"] = pd.to_datetime(
    employees["Last Login Time"], format="%H:%M %p"
).dt.time

employees.head()
```

```python
employees[employees["First Name"] == "Larry"]
```

```python
employees[employees["First Name"].duplicated()]
employees[employees["First Name"].duplicated(keep="first")]
employees[employees["First Name"].duplicated(keep="last")]
employees[employees["First Name"].duplicated(keep=False)]
```

- Use the tilde symbol (`~`) to invert a `Series's` values. True will become Falses, and Falses will become Trues.

```python
employees[~employees["First Name"].duplicated(keep=False)]
```

```python
employees[employees["First Name"] == "Angela"]
```

## The drop_duplicates Method
- The `drop_duplicates` method deletes rows with duplicate values.
- By default, it will remove a row if *all* of its values are shared with another row.
- The `subset` parameter configures the columns to look for duplicate values within.

```python
employees = pd.read_csv("employees.csv").astype(
    {"Senior Management": bool, "Gender": "category", "Team": "category"}
)

employees["Start Date"] = pd.to_datetime(employees["Start Date"], format="%m/%d/%Y")
employees["Last Login Time"] = pd.to_datetime(
    employees["Last Login Time"], format="%H:%M %p"
).dt.time

employees.head()
```

```python
employees.drop_duplicates("Team")
employees.drop_duplicates("Team", keep="first")
employees.drop_duplicates("Team", keep="last")
employees.drop_duplicates("Team", keep=False)

employees.drop_duplicates("First Name", keep=False)
```

- Pass a list to `subset` parameter to look for duplicates across multiple columns.

```python
employees.drop_duplicates(["Senior Management", "Team"]).sort_values("Team")

employees.drop_duplicates(["Senior Management", "Team"], keep="last").sort_values(
    "Team"
)
```

## The where Method
- Like the square brackets, the `where` method accepts a Boolean `Series` to filter the `DataFrame`.
- The resulting `DataFrame` will be of the same length as the original.
- Rows that do not satisfy the filtering criteria will have `NaN`s.
- The `where` method is ideal when you want to preserve the original `DataFrame` structure.

```python
employees = pd.read_csv("employees.csv").astype(
    {"Senior Management": bool, "Gender": "category", "Team": "category"}
)

employees["Start Date"] = pd.to_datetime(employees["Start Date"], format="%m/%d/%Y")
employees["Last Login Time"] = pd.to_datetime(
    employees["Last Login Time"], format="%H:%M %p"
).dt.time

employees.head()
```

```python
has_nice_salary = employees["Salary"] > 100000
```

```python
employees.where(has_nice_salary)
```

```python
employees["Salary"] = employees["Salary"].where(has_nice_salary)
```

```python
employees
```

## The query Method
- The `query` method filters a `DataFrame` using a string argument.
- Its argument is a more readable alternative to Boolean filtering.

```python
employees = pd.read_csv("employees.csv").astype(
    {"Senior Management": bool, "Gender": "category", "Team": "category"}
)

employees["Start Date"] = pd.to_datetime(employees["Start Date"], format="%m/%d/%Y")
employees["Last Login Time"] = pd.to_datetime(
    employees["Last Login Time"], format="%H:%M %p"
).dt.time

employees.head()
```

- Use `and`, `or`, and `not` (or `&`, `|`, `~`) within the expression string.

```python
employees.query("Salary > 100000")
```

```python
employees.query("Gender == 'Female' and Team == 'Marketing'")
employees.query("Gender == 'Female' or Team == 'Marketing'")
```

```python
employees.query("60000 <= Salary <= 70000")
```

- Use the `in` and `not in` operators to check inclusion/exclusion.

```python
employees.query("Team in ['Legal', 'Sales', 'Product']")

employees.query("Team not in ['Legal', 'Sales', 'Product']")
```

- Use the `@` prefix to reference local variables.

```python
min_salary = 130000
```

```python
employees.query("Salary > @min_salary")
```
